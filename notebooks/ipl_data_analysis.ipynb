```python
# Which batsman scored the most runs overall?
batsman_total_runs=df_x.groupby('batsman')['batsman_runs'].sum().sort_values(ascending=False)
print(batsman_total_runs,"\n")
# Find the batsman with the maximum runs
top_batsman = batsman_total_runs.idxmax()
top_batsman_runs = batsman_total_runs.max()
print("(top_batsman, top_batsman_runs):",top_batsman, top_batsman_runs)
```

```python
#Which bowler took the most wickets?
df_x[
    df_x['player_dismissed'].notna() & df_x['dismissal_kind'].notna()
]["bowler"].value_counts().sort_values(ascending=False)
```

```python
#What is the distribution of extras (wide, no-ball, leg-byes)?
extras_distribution = {
    'wide_runs': df_x['wide_runs'].sum(),
    'noball_runs': df_x['noball_runs'].sum(),
    'bye_runs': df_x['bye_runs'].sum(),
    'legbye_runs': df_x['legbye_runs'].sum()
}

extras_distribution
```

```python
# Which team scored the highest runs in a single match?
df_x.groupby(['match_id', 'batting_team'])['total_runs'].sum().sort_values(ascending=False)
```

```python
#Plot the trend of total runs scored per over in a match (visualization).

import matplotlib.pyplot as plt

runs_per_over = df_x.groupby(['match_id', 'over'])['total_runs'].sum().reset_index()

match_id_to_plot = 1
match_data = runs_per_over[runs_per_over['match_id'] == match_id_to_plot]

plt.figure(figsize=(10, 6))
plt.plot(match_data['over'], match_data['total_runs'], marker='o', linestyle='-', color='b')
plt.title(f"Trend of Total Runs Scored Per Over (Match ID: {match_id_to_plot})")
plt.xlabel("Over")
plt.ylabel("Total Runs")
plt.grid(True)
plt.show()
```

```python
#Visualization: Use Matplotlib or Seaborn to create the following visualizations:
 #Plot the top 5 teams with the most wins.
a=df_y["winner"].value_counts().sort_values(ascending=False).head()
print(a)
plt.figure(figsize=(12, 6))
plt.title("Top 5 teams with the most wins")
plt.xlabel("Teams")
plt.ylabel("No. of wins")
plt.plot(a,marker='o')
plt.show()
```

```python
#Bar chart of the top 5 batsmen with the highest runs
batsman_runs =df_x.groupby('batsman')['batsman_runs'].sum()

top_5_batsmen = batsman_runs.sort_values(ascending=False).head(5)

plt.figure(figsize=(10, 6))
top_5_batsmen.plot(kind='bar', color='skyblue')
plt.title('Top 5 Batsmen with Highest Runs')
plt.xlabel('Batsman')
plt.ylabel('Total Runs')
plt.show()
```

```python
#Distribution of winning margins (runs and wickets)
plt.figure(figsize=(10,6))
# Winning margin by runs
plt.subplot(1, 2, 1)
plt.hist(df_y['win_by_runs'], bins=20, color='blue', alpha=0.7, edgecolor='black')
plt.title('Distribution of Winning Margins (Runs)', fontsize=14)
plt.xlabel('Winning Margin (Runs)', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Winning margin by wickets
plt.subplot(1, 2, 2)
plt.hist(df_y['win_by_wickets'], bins=10, color='green', alpha=0.7, edgecolor='black')
plt.title('Distribution of Winning Margins (Wickets)', fontsize=14)
plt.xlabel('Winning Margin (Wickets)', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.show()
```

```python
#Line plot showing runs scored across overs in a specific match
runs_per_over = df_x.groupby(['match_id', 'over'])['total_runs'].sum().reset_index()

match_id_to_plot = 1
match_data = runs_per_over[runs_per_over['match_id'] == match_id_to_plot]

plt.figure(figsize=(10, 6))
plt.plot(match_data['over'], match_data['total_runs'], marker='o', linestyle='-', color='b')
plt.title(f"Runs Scored Across Overs(Match ID: {match_id_to_plot})")
plt.xlabel("Over")
plt.ylabel("Total Runs")
plt.grid(True)
plt.show()
```
