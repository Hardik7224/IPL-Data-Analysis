```python
# Data Cleaning and Preprocessing:
```

```python
# Identifying the null values
print(deliveries.isnull().sum())
print(matches.isnull().sum())
```

```python
# Handling the missing values
matches.fillna({'city': 'Unknown', 'winner': 'No Result'}, inplace=True)
deliveries.fillna({'batsman': 'Unknown', 'bowler': 'Unknown', 'player_dismissed': 'None'}, inplace=True)

print(deliveries.isnull().sum())
print(matches.isnull().sum())
```

```python
# Standardize names
team_name_mapping = {
    'Delhi Daredevils': 'Delhi Capitals',
    'Deccan Chargers': 'Sunrisers Hyderabad'
}


matches['team1'] = matches['team1'].replace(team_name_mapping)
matches['team2'] = matches['team2'].replace(team_name_mapping)
matches['winner'] = matches['winner'].replace(team_name_mapping)
deliveries['batting_team'] = deliveries['batting_team'].replace(team_name_mapping)
deliveries['bowling_team'] = deliveries['bowling_team'].replace(team_name_mapping)


print(matches['team1'].unique())
print(matches['team2'].unique())
```

```python
# Ensure Numerical Columns are Correctly Typed
print(matches.info())
print(deliveries.info())
```

```python
print(matches.columns)  
print(deliveries.columns)  

matches.rename(columns={'id': 'match_id'}, inplace=True)
print(matches.columns)

# Merge datasets on match_id
combined_data = pd.merge(deliveries, matches, on='match_id', how='inner')

# Verify merged dataset
print(combined_data.head())
print(combined_data.info())
```

```python
#Exploratory Data Analysis (EDA):
```

```python
#Team Win Percentages
team_wins = matches.groupby('winner').size()
print(team_wins)
total_matches = matches['team1'].value_counts() + matches['team2'].value_counts()
print(total_matches)

team_wins = team_wins.reindex(total_matches.index, fill_value=0)

win_percentage = (team_wins / total_matches) * 100

print(win_percentage)
```

```python
#Team with the Highest Winning Margin
highest_margin_runs = matches[matches['win_by_runs'] > 0].groupby('winner')['win_by_runs'].mean()
highest_margin_wickets = matches[matches['win_by_wickets'] > 0].groupby('winner')['win_by_wickets'].mean()

print(highest_margin_runs.idxmax(), highest_margin_wickets.idxmax())
```

```python
#Most Successful Captain
print(matches.columns)

successful_captains = matches[matches['toss_winner'] == matches['winner']]['toss_winner'].value_counts()

most_successful_captain = successful_captains.idxmax()
most_wins = successful_captains.max()

print(f"The most successful captain is {most_successful_captain} with {most_wins} matches won.")
```

```python
#Top 3 cities
import matplotlib.pyplot as plt

tied_matches = matches[matches['result'] == 'tie']

tied_by_city = tied_matches['city'].value_counts()

top_3_tied_cities = tied_by_city.head(3)

plt.figure(figsize=(8, 6))
top_3_tied_cities.plot(kind='bar', color=['skyblue', 'orange', 'green'])
plt.title('Top 3 Cities with the Most Tied Matches')
plt.xlabel('City')
plt.ylabel('Number of Tied Matches')
plt.show()

print(top_3_tied_cities)
```

```python
#Closest matches on average (smallest winning margins)
# Calculate average winning margins (runs and wickets) per season
average_margins = matches.groupby("season").agg({
    "win_by_runs": "mean",
    "win_by_wickets": "mean"
}).reset_index()

# Calculate a "closeness" metric (lower is closer matches)
# Using the sum of the averages as a single metric
average_margins["closeness_metric"] = average_margins["win_by_runs"] + average_margins["win_by_wickets"]

# Find the season with the smallest closeness metric
closest_season = average_margins.loc[average_margins["closeness_metric"].idxmin()]

# Display the results
print("Season with the closest matches:")
print(closest_season)

# Display all calculated averages for reference
print("\nAverage margins per season:")
print(average_margins)
```

```python
#Ball-Level Analysis 
```

```python
#Most Consistent Batsman
# Calculate total runs and matches played per batsman
batsman_stats = deliveries.groupby("batsman").agg({
    "batsman_runs": "sum",
    "match_id": pd.Series.nunique  # Unique matches played
}).reset_index()

# Calculate the average runs per match
batsman_stats["avg_runs_per_match"] = batsman_stats["batsman_runs"] / batsman_stats["match_id"]

# Identify the most consistent batsman
most_consistent_batsman = batsman_stats.sort_values("avg_runs_per_match", ascending=False).iloc[0]

# Display results
print("Most Consistent Batsman:")
print(most_consistent_batsman)

```

```python
#Top 5 Batsmen with the Most Boundaries
# Filter for boundaries
fours = deliveries[deliveries["batsman_runs"] == 4].groupby("batsman")["batsman_runs"].count()
sixes = deliveries[deliveries["batsman_runs"] == 6].groupby("batsman")["batsman_runs"].count()

# Combine into a single DataFrame
boundaries = pd.DataFrame({
    "fours": fours,
    "sixes": sixes
}).fillna(0)

# Total boundaries
boundaries["total_boundaries"] = boundaries["fours"] + boundaries["sixes"]

# Top 5 batsmen
top_5_boundaries = boundaries.sort_values("total_boundaries", ascending=False).head(5)

print("Top 5 Batsmen with Most Boundaries:")
print(top_5_boundaries)

```

```python
#Analyze Strike Rate in Powerplay
# Filter data for powerplay and death overs
powerplay = deliveries[deliveries["over"].between(1, 6)]
death_overs = deliveries[deliveries["over"].between(16, 20)]

# Calculate total runs and balls faced for both periods
strike_rate_powerplay = powerplay.groupby("batsman").agg({
    "batsman_runs": "sum",
    "ball": "count"
}).rename(columns={"batsman_runs": "runs", "ball": "balls"})
strike_rate_powerplay["strike_rate"] = (strike_rate_powerplay["runs"] / strike_rate_powerplay["balls"]) * 100

strike_rate_death = death_overs.groupby("batsman").agg({
    "batsman_runs": "sum",
    "ball": "count"
}).rename(columns={"batsman_runs": "runs", "ball": "balls"})
strike_rate_death["strike_rate"] = (strike_rate_death["runs"] / strike_rate_death["balls"]) * 100

# Top 5 players with highest strike rate in each period
top_5_powerplay = strike_rate_powerplay.sort_values("strike_rate", ascending=False).head(5)
top_5_death = strike_rate_death.sort_values("strike_rate", ascending=False).head(5)

print("Top 5 in Powerplay:")
print(top_5_powerplay)
print("\nTop 5 in Death Overs:")
print(top_5_death)
```

```python
#Most Economical Bowler
# Calculate total runs conceded and total balls bowled by each bowler
bowler_stats = deliveries.groupby("bowler").agg({
    "total_runs": "sum",
    "ball": "count"
}).reset_index()

# Convert balls into overs
bowler_stats["overs"] = bowler_stats["ball"] // 6 + (bowler_stats["ball"] % 6) / 6

# Filter bowlers who have bowled at least 100 overs
eligible_bowlers = bowler_stats[bowler_stats["overs"] >= 100]

# Calculate economy rate (runs per over)
eligible_bowlers["economy_rate"] = eligible_bowlers["total_runs"] / eligible_bowlers["overs"]

# Find the most economical bowler
most_economical_bowler = eligible_bowlers.sort_values("economy_rate").iloc[0]

print("Most Economical Bowler:")
print(most_economical_bowler)
```

```python
#Bowler with the Highest Dot-Ball Percentage
# Filter for dot balls (total_runs = 0)
dot_balls = deliveries[deliveries["total_runs"] == 0].groupby("bowler")["ball"].count()

# Calculate total balls bowled by each bowler
total_balls = deliveries.groupby("bowler")["ball"].count()

# Calculate dot-ball percentage
dot_ball_percentage = (dot_balls / total_balls) * 100

# Identify the bowler with the highest dot-ball percentage
highest_dot_ball_bowler = dot_ball_percentage.sort_values(ascending=False).iloc[:1]

print("Bowler with the Highest Dot-Ball Percentage:")
print(highest_dot_ball_bowler)
```

```python
#Run Rate Trends Across Overs
# Define the overs categories
def categorize_over(over):
    if 1 <= over <= 6:
        return "Powerplay"
    elif 7 <= over <= 15:
        return "Middle Overs"
    else:
        return "Death Overs"

# Add a column for over category
deliveries["over_category"] = deliveries["over"].apply(categorize_over)

# Calculate total runs and balls faced for each over category and inning
run_rate_trends = deliveries.groupby(["inning", "over_category"]).agg({
    "total_runs": "sum",
    "ball": "count"
}).reset_index()

# Calculate run rate (runs per over)
run_rate_trends["run_rate"] = run_rate_trends["total_runs"] / (run_rate_trends["ball"] / 6)

print("Run Rate Trends Across Overs:")
print(run_rate_trends)
```

```python
#Identifying Matches with Low Totals Successfully
# Filter matches with scores <150 and wins by runs (defended)
low_total_matches = matches[(matches["win_by_runs"] > 0) & (matches["win_by_runs"] < 150)]

# Filter for matches where the first innings score was <150
low_total_defenses = low_total_matches[low_total_matches["win_by_runs"] > 0]

# Display results
print(low_total_defenses[["match_id", "season", "team1", "team2", "win_by_runs", "winner"]])
```

```python
#Most Common Mode of Dismissal
# Count dismissals by dismissal_kind
dismissal_counts = deliveries["dismissal_kind"].value_counts()

# Identify the most common mode of dismissal
most_common_dismissal = dismissal_counts.idxmax()
most_common_dismissal_count = dismissal_counts.max()

print("Most Common Dismissal:")
print(most_common_dismissal, most_common_dismissal_count)
```

```python
#Bowler Dismissing a Specific Batsman the Most Times
# Filter only rows with valid player dismissals
valid_dismissals = deliveries[~deliveries["player_dismissed"].isna()]

# Count dismissals by bowler and batsman
bowler_batsman_dismissals = valid_dismissals.groupby(["bowler", "player_dismissed"]).size().reset_index(name="dismissal_count")

# Find the pair with the maximum dismissals
top_bowler_batsman_pair = bowler_batsman_dismissals.sort_values("dismissal_count", ascending=False).iloc[0]

print("Bowler dismissing a specific batsman the most:")
print(top_bowler_batsman_pair)
```

```python
#Heatmap of Dismissals by Over and Inning
import seaborn as sns
import matplotlib.pyplot as plt

# Filter only rows with valid dismissals
valid_dismissals_heatmap = deliveries[~deliveries["dismissal_kind"].isna()]

# Group by over and inning to count dismissals
dismissals_by_over_inning = valid_dismissals_heatmap.groupby(["inning", "over"]).size().unstack(fill_value=0)

# Plot heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(dismissals_by_over_inning, annot=True, fmt="d", cmap="coolwarm", cbar=True)
plt.title("Heatmap of Dismissals by Over and Inning")
plt.xlabel("Over")
plt.ylabel("Inning")
plt.show()
```

```python
#Win Percentages of All Teams
# Calculate win percentages for each team
team_matches = pd.concat([matches["team1"], matches["team2"]]).value_counts()
team_wins = matches["winner"].value_counts()
win_percentages = (team_wins / team_matches * 100).sort_values(ascending=False)

# Plot win percentages
plt.figure(figsize=(12, 6))
win_percentages.plot(kind="barh", color="skyblue", edgecolor="black")
plt.title("Win Percentages of All Teams", fontsize=16)
plt.xlabel("Win Percentage", fontsize=12)
plt.ylabel("Teams", fontsize=12)
plt.gca().invert_yaxis()
plt.grid(axis="x", linestyle="--", alpha=0.7)
plt.show()
```

```python
#Batsman Performance Metrics
# Top 10 Batsmen Based on Total Runs
top_10_batsmen_runs = deliveries.groupby("batsman")["batsman_runs"].sum().sort_values(ascending=False).head(10)

# Top 10 Batsmen Strike Rate
batsman_balls = deliveries.groupby("batsman")["ball"].count()
top_10_batsmen_strike_rate = ((top_10_batsmen_runs / batsman_balls) * 100).sort_values(ascending=False).head(10)

# Top 10 Batsmen Boundaries (Fours + Sixes)
batsman_fours = deliveries[deliveries["batsman_runs"] == 4].groupby("batsman")["batsman_runs"].count()
batsman_sixes = deliveries[deliveries["batsman_runs"] == 6].groupby("batsman")["batsman_runs"].count()
batsman_boundaries = (batsman_fours + batsman_sixes).sort_values(ascending=False).head(10)

# Plot results
fig, axes = plt.subplots(1, 3, figsize=(18, 6), sharey=True)
top_10_batsmen_runs.plot(kind="bar", ax=axes[0], color="coral", edgecolor="black", title="Top 10 Batsmen - Total Runs")
top_10_batsmen_strike_rate.plot(kind="bar", ax=axes[1], color="orange", edgecolor="black", title="Top 10 Batsmen - Strike Rate")
batsman_boundaries.plot(kind="bar", ax=axes[2], color="gold", edgecolor="black", title="Top 10 Batsmen - Boundaries")

# Formatting
for ax in axes:
    ax.set_xlabel("Batsman", fontsize=12)
    ax.set_ylabel("Count", fontsize=12)
    ax.grid(axis="y", linestyle="--", alpha=0.7)
fig.suptitle("Batsman Performance Metrics", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
#Scatter Plot: Economy Rate vs. Dot-Ball Percentage
# Calculate dot-ball percentage and economy rate for bowlers
dot_balls_bowler = deliveries[deliveries["total_runs"] == 0].groupby("bowler")["ball"].count()
total_balls_bowler = deliveries.groupby("bowler")["ball"].count()
total_runs_bowler = deliveries.groupby("bowler")["total_runs"].sum()

# Compute metrics
bowler_metrics = pd.DataFrame({
    "dot_ball_percentage": (dot_balls_bowler / total_balls_bowler) * 100,
    "economy_rate": total_runs_bowler / (total_balls_bowler / 6)
}).dropna()

# Top 10 bowlers based on dot-ball percentage
top_10_bowlers = bowler_metrics.sort_values("dot_ball_percentage", ascending=False).head(10)

# Scatter plot
plt.figure(figsize=(10, 6))
plt.scatter(
    top_10_bowlers["economy_rate"],
    top_10_bowlers["dot_ball_percentage"],
    color="blue", edgecolor="black", s=100
)
plt.title("Economy Rate vs. Dot-Ball Percentage (Top 10 Bowlers)", fontsize=16)
plt.xlabel("Economy Rate", fontsize=12)
plt.ylabel("Dot-Ball Percentage (%)", fontsize=12)
plt.grid(alpha=0.7)
plt.show()
```

```python
#1. Head-to-Head Team Analysis
"""This compares two teams (e.g., CSK vs. MI) based on:
-Total Wins
-Average Scores
-Win Margins"""

# Define the two teams
team_a = "Chennai Super Kings"
team_b = "Mumbai Indians"

# Head-to-head total wins
head_to_head_wins = matches[matches["team1"].isin([team_a, team_b]) & 
                               matches["team2"].isin([team_a, team_b])]["winner"].value_counts()

# Head-to-head average scores
head_to_head_scores = deliveries[deliveries["batting_team"].isin([team_a, team_b]) &
                                    deliveries["bowling_team"].isin([team_a, team_b])].groupby("batting_team")["total_runs"].mean()

# Head-to-head win margins
head_to_head_margins = matches[matches["winner"].isin([team_a, team_b])].groupby("winner").agg({
    "win_by_runs": "mean",
    "win_by_wickets": "mean"
})

# Display results
print("Head-to-Head Wins:")
print(head_to_head_wins)
print("\nAverage Scores:")
print(head_to_head_scores)
print("\nWin Margins:")
print(head_to_head_margins)
```

```python
#2. Impact Players
"""Identify players who contributed most to victories based on:

-Highest Individual Scores
-Wickets Taken in Low-Scoring Matches"""

# Highest individual scores in matches won
winning_match_ids = matches[matches["winner"].notna()]["match_id"]
impact_batsmen = deliveries[deliveries["match_id"].isin(winning_match_ids)].groupby(["match_id", "batsman"]).agg({
    "batsman_runs": "sum"
}).reset_index().sort_values("batsman_runs", ascending=False).head(10)

# Most wickets in low-scoring matches
low_scoring_matches = matches[(matches["win_by_runs"] > 0) & (matches["win_by_runs"] < 150)]["match_id"]
impact_bowlers = deliveries[deliveries["match_id"].isin(low_scoring_matches)].groupby(["match_id", "bowler"])["player_dismissed"].count().reset_index().sort_values("player_dismissed", ascending=False).head(10)

# Display results
print("Impact Batsmen (Top 10):")
print(impact_batsmen)
print("\nImpact Bowlers (Top 10):")
print(impact_bowlers)
```

```python
#Win Prediction Analysis
"""Analyze trends in toss decisions and their impact on match results:
-Does choosing to field first increase winning probability?"""

# Count wins based on toss decisions
toss_decision_wins = matches[matches["toss_winner"] == matches["winner"]].groupby("toss_decision").size()

# Toss win correlation with match win
toss_win_correlation = matches["toss_winner"].eq(matches["winner"]).mean() * 100

# Display results
print("Wins Based on Toss Decision:")
print(toss_decision_wins)
print("\nToss-Win Match-Win Correlation:")
print(f"{toss_win_correlation:.2f}%")
```
